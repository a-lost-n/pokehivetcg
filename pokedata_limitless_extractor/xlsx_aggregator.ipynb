```python
import pandas as pd
import os
```

```python
decks_df = pd.DataFrame()
matchups_df = pd.DataFrame()
for tournament in os.listdir('datasets'):
    aux_deck_df = pd.read_excel(f'datasets/{tournament}', sheet_name='decks')
    aux_deck_df['Tournament'] = tournament[:-5]
    decks_df = pd.concat([decks_df, aux_deck_df])

    aux_matchups_df = pd.read_excel(f'datasets/{tournament}', sheet_name='matchups')
    matchups_df = pd.concat([matchups_df, aux_matchups_df])

matchups_df = matchups_df.groupby(['Deck', 'Opposing Deck'], as_index=False).sum()
```

```python
with pd.ExcelWriter('temporal_forces.xlsx') as writer:
    decks_df.to_excel(writer, sheet_name='decks', index=False)
    matchups_df.to_excel(writer, sheet_name='matchups', index=False)
```
