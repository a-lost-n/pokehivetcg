# Image cropper
Takes screenshots of the PDF generated by MS Power BI and cropps them to the correct dimension apt for posting on X

Run to clean the report folder first: 

rm -f projects/pokehivetcg/image_cropper/report/*.png; rm -f projects/pokehivetcg/image_cropper/uncropped_report/*.png


```python
from PIL import Image
import os

origin = 'uncropped_report'
destiny = 'report'
```

## Resize to 16:9

```python
# 12 decks
for img_name in os.listdir(origin):
    img = Image.open("{}/{}".format(origin, img_name))
    width, height = img.size
    img = img.crop((475,34,width-477,height-39))
    img.save('{}/{}'.format(destiny, img_name))
```

## Resize to 8:9

```python
for img_name in ['1', '2']:
    img = Image.open("{}/{}.png".format(origin, img_name))
    width, height = img.size
    img = img.crop((690,50,width-690,height-61))
    img.save('{}/{}.png'.format(destiny, img_name))
```
