# Image cropper
Takes screenshots of the PDF generated by MS Power BI and cropps them to the correct dimension apt for posting on X

Run to clean the report folder first: 

rm -f image_cropper/report/*.PNG


```python
from PIL import Image
import os
```

## Resize to 16:9

```python
for img_name in os.listdir('report'):
    img = Image.open("report/{}".format(img_name))
    width, height = img.size
    img = img.crop((98,50,width-100,height-78))
    img.save('cropped_report/{}'.format(img_name))
```

## Resize to 8:9

```python
img = Image.open("report/1.PNG")
width, height = img.size
img = img.crop((687,50,width-690,height-55))
img.save('cropped_report/1.PNG')
img = Image.open("report/2.PNG")
width, height = img.size
img = img.crop((687,50,width-690,height-55))
img.save('cropped_report/2.PNG')
```

## Fix issues on 16:9 crops

```python
def custom_crop(img_name, left=0, top=0, right=0, bottom=0):
    img = Image.open("report/{}".format(img_name))
    width, height = img.size
    img = img.crop((73+left,37+top,width-74+right,height-45+bottom))
    img.save('cropped_report/{}'.format(img_name))

custom_crop('8.PNG',top=2)
custom_crop('10.PNG',top=2)
custom_crop('11.PNG',top=2)
```
