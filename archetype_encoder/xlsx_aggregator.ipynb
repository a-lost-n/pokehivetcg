```python
import pandas as pd
import os
```

```python
dir = 'merge_datasets'
```

```python
decks_df = pd.DataFrame()
pairings_df = pd.DataFrame()
for tournament in os.listdir(dir):
    aux_deck_df = pd.read_excel(f'{dir}/{tournament}', sheet_name='decks')
    aux_deck_df['Tournament'] = tournament[:-5]
    decks_df = pd.concat([decks_df, aux_deck_df])

    aux_pairings_df = pd.read_excel(f'{dir}/{tournament}', sheet_name='pairings')
    pairings_df = pd.concat([pairings_df, aux_pairings_df])

# p_df = pairings_df.groupby(['Deck', 'Opposing Deck'], as_index=False).sum()
```

```python
with pd.ExcelWriter('merged_dataset.xlsx') as writer:
    decks_df.to_excel(writer, sheet_name='decks', index=False)
    pairings_df.to_excel(writer, sheet_name='pairings', index=False)
```
