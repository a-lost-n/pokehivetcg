```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from variables import TOUR_NAME, LIMITLESS_TOUR_ENDPOINT
```

```python
output = pd.read_excel(f'standings/{TOUR_NAME}_standings.xlsx', sheet_name="masters")
```

```python
def parse_limitless(endpoint):
    with requests.get(endpoint) as html_content:
        soup = BeautifulSoup(html_content.text, 'html.parser')
        table = soup.find('table', class_='data-table striped')
        data_tooltips = []
        for row in table.find_all('tr')[1:]:
            deck = row.find('span')['data-tooltip']
            data_tooltips.append(deck if deck else 'Other')
        return data_tooltips

decks = parse_limitless(LIMITLESS_TOUR_ENDPOINT)
for i in range(len(decks)):
    if output['Deck'][i] == "Other":
        print(f'Other at row {i}')
    elif output['Deck'][i] != decks[i]:
        print(f'Error at row {i}')
```
